```python
from datetime import date

event_title = input("Please enter the event title: ")
year = int(input("Enter the year: "))
month = int(input("Enter the month: "))
day = int(input("Enter the day: "))

event_date = date(year, month, day)

month_name = event_date.strftime("%B")
weekday_name = event_date.strftime("%A")

print("Event: ", event_title.upper())

print(f'Date: {month_name} {event_date.day}, {event_date.year}')

print("Year:", year)

print(f'Month: {month_name}' )

print("Day: ", day)

print("Weekday: ",weekday_name )
```
